```python
import pandas as pd
import os
import re
```

```python
path = "metrics/"
```

```python
csv_files = [f for f in os.listdir(path) if f.endswith(".csv") and "Grid" not in f]

dfs = []
```

```python
def extract_representation(filename):
    name = filename.replace(".csv", "")
    return name.replace("antiviral_homology_90_", "").replace("antiviral_homology_90", "")

# Leer y etiquetar
for file in csv_files:
    df = pd.read_csv(os.path.join(path, file))
    df["representation"] = extract_representation(file)
    dfs.append(df)
```

```python
combined_df = pd.concat(dfs, ignore_index=True)
combined_df
```

```python
not_trained=combined_df[combined_df["Dataset"] != "Train"]
```

```python
not_trained
```

```python
mean_grouped = not_trained.groupby(["Dataset", "model", "sampling","representation"]).mean(numeric_only=True)
mean_grouped = mean_grouped.reset_index()
```

```python
mean_grouped
```

```python
mean_grouped.to_csv("resume_metrics.csv", index=False)
```
